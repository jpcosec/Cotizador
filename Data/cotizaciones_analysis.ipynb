# Cotizaciones Analysis

Notebook para explorar `Data_Historica.csv`.

```python
import pandas as pd

csv_path = 'Data_Historica.csv'
df = pd.read_csv(csv_path)
df.head()
```

```python
print('Rows:', len(df))
print('Unique files:', df['Archivo'].nunique())
print('Columns:', list(df.columns))
```

```python
# Ejemplo de filtro por empresa
empresa = 'Johnsons -SOFOFA.'
df[df['Empresa'].fillna('').str.contains(empresa, case=False, na=False)].head(20)
```

```python
# Top servicios más repetidos
df['Servicio'].value_counts().head(20)
```

```python
import matplotlib.pyplot as plt

# Parseo robusto de fecha y creación de columna year
df['Fecha_dt'] = pd.to_datetime(df['Fecha'], errors='coerce', dayfirst=True)
df['year'] = df['Fecha_dt'].dt.year

# Fallback por regex cuando la fecha no parsea bien
mask_year_na = df['year'].isna()
year_from_text = pd.to_numeric(
    df.loc[mask_year_na, 'Fecha'].astype(str).str.extract(r'((?:19|20)\\d{2})', expand=False),
    errors='coerce'
)
df.loc[mask_year_na, 'year'] = year_from_text
df['year'] = df['year'].astype('Int64')

print('Years detectados:', df['year'].dropna().sort_values().unique()[:5], '...')
print('Rango años:', int(df['year'].min()), '-', int(df['year'].max()))
print('Filas sin year:', int(df['year'].isna().sum()))
```

```python
# Pivot: Servicio x year (conteo de registros)
svc_year = (
    df.dropna(subset=['Servicio', 'year'])
      .pivot_table(index='Servicio', columns='year', values='Archivo', aggfunc='count', fill_value=0)
      .sort_index()
)

svc_year.head(20)
```

```python
# Visual rápida: top 25 servicios por total histórico
top_services = svc_year.sum(axis=1).sort_values(ascending=False).head(25).index
plot_df = svc_year.loc[top_services]

ax = plot_df.plot(kind='bar', stacked=True, figsize=(16, 7), colormap='tab20')
ax.set_title('Servicios por año (conteo) - Top 25')
ax.set_xlabel('Servicio')
ax.set_ylabel('Cantidad')
plt.xticks(rotation=75, ha='right')
plt.tight_layout()
plt.show()
```

```python
# Chequeo de posibles renombres/variantes por normalización simple
tmp = df.dropna(subset=['Servicio']).copy()
tmp['Servicio_norm'] = tmp['Servicio'].astype(str).str.strip().str.lower()

possible_variants = (
    tmp.groupby('Servicio_norm')['Servicio']
       .agg(lambda s: sorted(set(s)))
       .reset_index(name='variantes')
 )
possible_variants['n_variantes'] = possible_variants['variantes'].str.len()

possible_variants[possible_variants['n_variantes'] > 1].sort_values('n_variantes', ascending=False).head(30)
```
